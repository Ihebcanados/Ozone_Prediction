```python
import pandas as pd
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from numpy import sqrt
from sklearn.metrics import mean_squared_error,r2_score




df=pd.read_excel('ozone.xlsx')
```

```python
df.describe()
```

```python
df.shape
```

```python
df_x=df[['T9','T12','T15','Ne9','Ne12','Ne15','Vx9','Vx12','Vx15',]]
df_y=df[['MaxO3']]
for col in df_x.columns:
  x=df_x[[col]].values.astype(float)
  stand_n=preprocessing.StandardScaler()
  df_x[[col]]=stand_n.fit_transform(df_x[[col]])

print(df_x.head())



```

```python
X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.2)
print("Ensemble d'entraînement :", X_train.shape, y_train.shape)
print("Ensemble de test :", X_test.shape, y_test.shape)

reg_lin=LinearRegression()

reg_lin.fit(X_train,y_train)

train_pred=reg_lin.predict(X_train)
test_pred=reg_lin.predict(X_test)
print(train_pred)
print(test_pred)

```

```python
rmse=sqrt(mean_squared_error(y_train,train_pred))
print(rmse)

rmse1=sqrt(mean_squared_error(y_test,test_pred))
print(rmse1)

r2=r2_score(y_train,train_pred)
r21=r2_score(y_test,test_pred)
print(r2)
print(r21)
```
